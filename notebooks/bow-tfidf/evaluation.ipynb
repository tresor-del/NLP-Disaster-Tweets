```python
import numpy as np
import joblib

y_test = np.load("../data/processed/y_test.npy")
y_pred = np.load("../data/predictions/y_pred.npy")

model = joblib.load("../model/xgboost_model.pkl")
```

### Confusion Matrix

```python
import seaborn as sns
import matplotlib.pyplot as plt
from sklearn.metrics import confusion_matrix

cm = confusion_matrix(y_pred=y_pred, y_true=y_test)
labels = [[f"TN: {cm[0,0]}", f"FP: {cm[0,1]}"], [f"FN: {cm[1,0]}", f"TP: {cm[1,1]}"]]

sns.heatmap(cm, annot=labels, fmt='', cmap='Blues', xticklabels=['Good', 'Disaster'], yticklabels=['Good', 'Disaster'], cbar=False)
plt.title('Disaster Tweets Confusion Matrix')
plt.show()
```

```python
from sklearn.metrics import accuracy_score

test_accuracy = accuracy_score(y_test, y_pred)
print(test_accuracy)
```

```python
from sklearn.metrics import precision_score, recall_score, f1_score

print("Precision:", precision_score(y_test, y_pred))
print("Recall:   ", recall_score(y_test, y_pred))
print("F1 Score: ", f1_score(y_test, y_pred))
```
