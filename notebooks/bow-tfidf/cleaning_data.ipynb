# Cleaning the data

```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv("../../data/raw/train.csv", encoding="latin-1")
display(df.shape)
display(df.head())
```

```python
df = df[["text", "target"]]
df
```

```python
# Missing Data
df.isnull().sum()
```

```python
# Duplicated Data
df.duplicated().sum()
```

```python
df.drop_duplicates(keep="first")
print("Shape: ", df.shape)
print("Class Distribution:", df["target"].value_counts())
```

```python
df.loc[:, "text"] = df["text"].str.lower()
```

```python
import re

df.loc[:, "text"] = df["text"].apply(lambda x: re.sub(r'[^\w\s]', '', x))
```

```python
df.head()
```

### tokenization

```python
import nltk
from nltk.tokenize import word_tokenize

nltk.download("punkt_tab")

df.loc[:, "token"] = df.loc[:, "text"].apply(word_tokenize)

df.head()
```

### stopwords

```python
import nltk
from nltk.corpus import stopwords

nltk.download("stopwords")

stop_words = set(stopwords.words('english'))

df.loc[:, "token"] = df.loc[:, "token"].apply(lambda x: [word for word in x if word not in stop_words])

df.head()
```

```python
df.loc[:, "processed_text"] = df["token"].apply(lambda x: " ".join(x))
df.head()
```

```python

df.to_csv("../../data/processed/cleaned_data.csv", index=False)

print("Cleaned data saved as 'cleaned_data.csv'")
```
